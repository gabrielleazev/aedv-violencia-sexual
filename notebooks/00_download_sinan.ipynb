```python
import pysus
import pandas as pd
import gc

anos = range(2015, 2025)
partes = []

for ano in anos:
    try:
        df = pysus.ftp.sinan(disease="VIOL", year=ano, as_dataframe=True)
        sexual = df[(df["VIOL_SEXU"].astype(str) == "1") & (df["CS_SEXO"] == "F")].copy()
        sexual["ANO_ARQUIVO"] = ano
        partes.append(sexual)
        print(f"{ano}: {len(df):>7} notificações | {len(sexual):>6} violência sexual F")
        del df
        gc.collect()
    except Exception as e:
        print(f"{ano}: ERRO -> {e}")

sinan = pd.concat(partes, ignore_index=True)
print("\nTotal:", sinan.shape)
print(sinan["ANO_ARQUIVO"].value_counts().sort_index())
```

```python
sinan = sinan.astype(str)
sinan.to_parquet("sinan_violencia_sexual_f_2015_2024.parquet", index=False)

from google.colab import files
files.download("sinan_violencia_sexual_f_2015_2024.parquet")
```
